## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Section 3: Quick look at the data (2-3 real numbers)

Below, we load the anonymized dataset to evaluate key baseline numbers supporting Lane 1:

1. Baseline Declining Rate: The overall proportion of pages in the dataset experiencing a downward trend.

2. Stale & Visible Pages: The number of pages with significant exposure (impressions_90d >= 500) that have not been updated recently (days_since_last_update >= 180).

3. High-Impression Declining Pages: The fraction of top-traffic pages (impressions_90d > 5000) that are currently decaying.

```python
import os
import pandas as pd

# Pathway logic with automatic download fallback for Colab
data_path = "data/raw/content_refresh_anonymized.csv"

if not os.path.exists(data_path):
    if os.path.exists("../../data/raw/content_refresh_anonymized.csv"):
        data_path = "../../data/raw/content_refresh_anonymized.csv"
    else:
        # Download directly from repo if running in Google Colab environment
        url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
        print("Downloading dataset from GitHub repository...")
        df = pd.read_csv(url)
        data_path = None

if data_path:
    df = pd.read_csv(data_path)

df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

# Real Number 1: Baseline Declining Rate
overall_declining_rate = df["is_declining_label"].mean()

# Real Number 2: Stale & Visible Count (impressions >= 500 & days_since_last_update >= 180)
stale_visible = df[(df["impressions_90d"] >= 500) & (df["days_since_last_update"] >= 180)]
stale_visible_count = len(stale_visible)

# Real Number 3: High Exposure Declining Rate (impressions > 5000)
high_imp = df[df["impressions_90d"] > 5000]
high_imp_decay_rate = high_imp["is_declining_label"].mean()

print(f"1. Total Pages: {len(df):,} | Baseline Declining Rate: {overall_declining_rate:.1%}")
print(f"2. Stale & Visible Pages (>=500 imps, >=180 days old): {stale_visible_count:,}")
print(f"3. Declining Rate among High-Exposure Pages (>5k imps): {high_imp_decay_rate:.1%}")
```
